```python
# Install packages
!pip install streamlit pyngrok opencv-python pillow -q

import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

print("CuraMind AI - Medical Image Diagnostic Demo")

# Upload medical image (Xray/MRI sample)
from google.colab import files
uploaded = files.upload()

# Read image
for file in uploaded.keys():
    img = Image.open(file)
    img_array=np.array(img)

# Convert to grayscale
gray=cv2.cvtColor(img_array,cv2.COLOR_BGR2GRAY)

# Simple anomaly highlighting using edge detection
edges=cv2.Canny(gray,50,150)

# Simulated diagnostic prediction
mean_intensity=np.mean(gray)

if mean_intensity>120:
    diagnosis="Potential Abnormality Detected"
    risk="High"
else:
    diagnosis="Normal Scan"
    risk="Low"

# Show results
print("\nDiagnostic Report")
print("----------------------")
print("Diagnosis:",diagnosis)
print("Risk Level:",risk)

# Display images
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(img_array)
plt.title("Uploaded Medical Scan")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(edges,cmap='gray')
plt.title("Detected Anomaly Regions")
plt.axis("off")

plt.show()

# Patient Record Example
patient_record={
    "Patient":"Demo Patient",
    "Scan":"MRI/X-ray",
    "Diagnosis":diagnosis,
    "Risk":risk
}

print("\nElectronic Medical Record")
print(patient_record)
```
